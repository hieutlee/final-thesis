```python
from ultralytics import YOLO
import cv2
import os

# Load trained YOLOv8 model
model = YOLO('C:/Users/hieut/FlickrLogos-v2/runs/detect/logo_training_v2/weights/best.pt')

# Input video
video_path = 'C:/Users/hieut/FlickrLogos-v2/Test Videos/testvideo.mp4'
cap = cv2.VideoCapture(video_path)

# Get video properties
fps = cap.get(cv2.CAP_PROP_FPS)
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fourcc = cv2.VideoWriter_fourcc(*'mp4v')

# Output path
output_path = 'C:/Users/hieut/FlickrLogos-v2/Test Videos/testvideo_output.mp4'
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

frame_count = 0

# Inference loop
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Inference on the current frame
    results = model(frame, verbose=False)

    # Draw results on the frame
    annotated_frame = results[0].plot()

    # Write the frame to output
    out.write(annotated_frame)

    # (Optional) Display while processing
    # cv2.imshow('Detection', annotated_frame)
    # if cv2.waitKey(1) & 0xFF == ord('q'):
    #     break

    frame_count += 1
    if frame_count % 30 == 0:
        print(f"Processed {frame_count} frames...")

# Cleanup
cap.release()
out.release()
cv2.destroyAllWindows()
print(f"Detection complete. Annotated video saved to: {output_path}")
```
