```python
# 1. Import necessary libraries
from ultralytics import YOLO  # Import YOLO
import os

# 2. Set the path to the dataset and YAML file
yaml_filename = 'C:/Users/hieut/FlickrLogos-v2/data.yaml'  # Replace with your actual path

# 3. Load YOLO model
model = YOLO('yolov8s.pt')  # GPU is now RTX 4080 Laptop GPU rather than RTX3080 Desktop. 

# 4. Train the model
model.train(
    data=yaml_filename,      # Path to the data.yaml file
    epochs=100,              # Set number of epochs (adjust based on your preference)
    batch=16,                # Batch size (adjust based on your preference)
    imgsz=640,               # Image size (resize images to 640x640, adjust if needed)
    name='logo_training_v2', # Name for this training run
    lr0=0.001,
    augment=True,  # Enable data augmentation
    patience=10,             # Early stopping patience
    device=0,                # GPU device index, use 0 for the primary GPU
    val=True                 # Enable validation during training
)
```
